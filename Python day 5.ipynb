```python
import pandas as pd
import numpy as np

# Creating a DataFrame with None and np.nan (null/missing values)
data = {
    "Name": ["Alice", "Bob", None, "David", "Eva"],
    "Age": [25, np.nan, 30, 22, np.nan],
    "City": ["Lucknow", "Delhi", "Mumbai", None, "Kanpur"],
    "Score": [85.5, 90.0, np.nan, 88.0, None]
}

df = pd.DataFrame(data)

# Display the DataFrame
print("--- DataFrame ---")
print(df)

# Check for missing/null values (returns True where values are missing)
print("\n--- Missing Values Mask ---")
print(df.isna())

# Count missing values per column
print("\n--- Missing Values Count Per Column ---")
print(df.isna().sum())
```

```python
df["Score"]=df["Score"].fillna(df["Score"].median())
print(df)
```
