```python
class Student:
    def __init__(self, name, roll_no, course, branch):
        self.name = name
        self.roll = roll_no
        self.course = course
        self.branch = branch

    def display(self):
        print("Welcome to Python")
        print(f"Name: {self.name}")
        print(f"Roll no.: {self.roll}")
        print(f"Course: {self.course}")
        print(f"Branch: {self.branch}")
        print("-" * 20) 
        

s1 = Student("Ayukt Yadav", 26, "B.Tech", "Information Technology")
s2 = Student("Rahul Sharma", 30, "B.Tech", "Computer Science")

s1, s2 = s2, s1

s1.display()
s2.display()
```
