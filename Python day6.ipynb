```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import make_interp_spline

# Original data points
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 25, 15, 30, 20])

# Generate 300 evenly spaced points for a smooth curve
x_smooth = np.linspace(x.min(), x.max(), 300)
spl = make_interp_spline(x, y, k=3)  # k=3 for cubic spline
y_smooth = spl(x_smooth)

# Plot
plt.plot(x_smooth, y_smooth, label="Smooth Curve", color="blue", linewidth=2)
plt.scatter(x, y, color="red", zorder=5, label="Data Points")  # Original points

plt.title("Smooth Curve through Points")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Star ke 5 points ke liye angles
angles = np.linspace(0, 2 * np.pi, 6)
# Standard star shape ke liye angles sequence
star_angles = [angles[0], angles[2], angles[4], angles[1], angles[3], angles[0]]

r = [1] * 6  # Radius

# Polar plot me star draw karna
plt.figure(figsize=(5, 5))
plt.subplot(projection="polar")
plt.plot(star_angles, r, color="gold", linewidth=3)
plt.fill(star_angles, r, color="yellow", alpha=0.5)  # Star me color bharne ke liye
plt.axis("off")  # Grid aur axes chhipane ke liye
plt.title("Star Shape", fontsize=14)
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Naya plot aur axes set karein
fig, ax = plt.subplots(figsize=(6, 6))

# ---- Ghar ki diwarein (Rectangle) ----
# origin (0.3, 0.1), width 0.4, height 0.4
base = patches.Rectangle((0.3, 0.1), 0.4, 0.4, linewidth=2, edgecolor='black', facecolor='tan', label='Diwarein')
ax.add_patch(base)

# ---- Chhat (Triangle) ----
# Points: (0.3, 0.5), (0.7, 0.5), (0.5, 0.8)
roof = patches.Polygon([[0.3, 0.5], [0.7, 0.5], [0.5, 0.8]], closed=True, linewidth=2, edgecolor='brown', facecolor='brown', label='Chhat')
ax.add_patch(roof)

# ---- Darwaza (Rectangle) ----
# origin (0.45, 0.1), width 0.1, height 0.2
door = patches.Rectangle((0.45, 0.1), 0.1, 0.2, linewidth=2, edgecolor='black', facecolor='saddlebrown', label='Darwaza')
ax.add_patch(door)

# ---- Khidki (Circle) ----
# Center (0.5, 0.65), radius 0.04
window = patches.Circle((0.5, 0.65), 0.04, linewidth=2, edgecolor='black', facecolor='skyblue', label='Khidki')
ax.add_patch(window)


# Axes boundaries aur background set karein
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect('equal')  # Ghar disproportionate na dikhe
ax.axis('off')  # Grid aur axes ko chhipane ke liye

# Legend dikhane ke liye
ax.legend(loc='upper right')

plt.title("A Simple House Draw using Matplotlib")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


def draw_flower(petals=6, petal_color="pink", center_color="gold"):
    # Generate angles from 0 to 2*pi
    theta = np.linspace(0, 2 * np.pi, 1000)

    # Polar equation for petals: r = 1 + A * sin(n * theta)
    r = 1 + 0.5 * np.sin(petals * theta)

    # Create polar plot
    plt.figure(figsize=(5, 5))
    ax = plt.subplot(111, projection="polar")

    # Plot petal outline and fill with pink color
    ax.plot(theta, r, color="deeppink", linewidth=2)
    ax.fill(theta, r, color=petal_color, alpha=0.8)

    # Draw the center circle of the flower
    center_r = np.full_like(theta, 0.25)
    ax.fill(theta, center_r, color=center_color)

    # Turn off axes and background grid
    ax.axis("off")
    plt.title("Pink Flower", fontsize=14, pad=15)
    plt.show()


# Call the function to draw a 6-petaled pink flower
draw_flower(petals=6, petal_color="pink")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# 1. Outer Rounded Rectangle (Main Body)
outer_box = patches.FancyBboxPatch(
    (0.1, 0.1), 0.8, 0.8,
    boxstyle="round,pad=0.0,rounding_size=0.2",
    ec="#E1306C", lw=12, fc="none"
)
ax.add_patch(outer_box)

# 2. Outer Camera Lens (Circle)
lens_outer = patches.Circle((0.5, 0.5), 0.2, ec="#E1306C", lw=10, fc="none")
ax.add_patch(lens_outer)

# 3. Camera Flash (Dot)
flash = patches.Circle((0.7, 0.7), 0.035, fc="#E1306C")
ax.add_patch(flash)

# Set equal scaling and axis limits
ax.set_aspect('equal')
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

# Hide axes tick marks and borders
plt.axis("off")

# Display the plot
plt.show()
```

```python
import matplotlib.patches as patches
import matplotlib.pyplot as plt
import numpy as np


def draw_solar_system():
    fig, ax = plt.subplots(figsize=(10, 10), facecolor="black")
    ax.set_facecolor("black")

    # ---- Planet Data: (Name, Orbit Radius, Planet Size, Color) ----
    planets = [
        ("Mercury", 0.8, 0.03, "gray"),
        ("Venus", 1.3, 0.05, "goldenrod"),
        ("Earth", 1.8, 0.06, "dodgerblue"),
        ("Mars", 2.3, 0.04, "indianred"),
        ("Jupiter", 3.2, 0.14, "sandybrown"),
        ("Saturn", 4.1, 0.11, "khaki"),
        ("Uranus", 4.9, 0.08, "mediumturquoise"),
        ("Neptune", 5.6, 0.08, "royalblue"),
    ]

    # ---- Draw Sun ----
    sun = patches.Circle((0, 0), 0.35, color="gold", zorder=3)
    ax.add_patch(sun)
    ax.text(
        0,
        0,
        "Sun",
        color="black",
        fontsize=9,
        fontweight="bold",
        ha="center",
        va="center",
        zorder=4,
    )

    # Angles for placing planets around their orbit
    np.random.seed(15)
    angles = np.random.uniform(0, 2 * np.pi, len(planets))

    # ---- Draw Orbits and Planets ----
    theta = np.linspace(0, 2 * np.pi, 300)

    for i, (name, r, size, color) in enumerate(planets):
        # 1. Draw Orbit Line
        ax.plot(
            r * np.cos(theta),
            r * np.sin(theta),
            color="gray",
            linestyle="--",
            linewidth=0.8,
            alpha=0.5,
        )

        # 2. Compute Planet Coordinates
        angle = angles[i]
        x_p = r * np.cos(angle)
        y_p = r * np.sin(angle)

        # 3. Special Case: Saturn's Rings
        if name == "Saturn":
            ring = patches.Ellipse(
                (x_p, y_p),
                width=size * 3,
                height=size * 0.8,
                angle=20,
                edgecolor="bisque",
                facecolor="none",
                linewidth=1.5,
                zorder=2,
            )
            ax.add_patch(ring)

        # 4. Draw Planet Body
        planet_circle = patches.Circle(
            (x_p, y_p), size, color=color, zorder=3
        )
        ax.add_patch(planet_circle)

        # 5. Label Planet
        ax.text(
            x_p,
            y_p + size + 0.1,
            name,
            color="white",
            fontsize=8,
            ha="center",
            va="bottom",
        )

    # Formatting Plot
    ax.set_xlim(-6.2, 6.2)
    ax.set_ylim(-6.2, 6.2)
    ax.set_aspect("equal")
    ax.axis("off")

    plt.title("Solar System", color="white", fontsize=16, pad=20)
    plt.tight_layout()
    plt.show()


# Call the function to draw the solar system
draw_solar_system()
```
